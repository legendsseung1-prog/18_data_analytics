```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지우"],
    "나이": [20, 21, 22, 20],
    "성별": ["남", "여", "남", "여"],
    "수학": [90, 85, 70, 100],
    "영어": [85, 95, 75, 90]
}
```

```python
df = pd.DataFrame(data)

print(df)
```

```python
print(df.head(2))
```

```python
df.info()
```

```python
df.set_index("이름", inplace=True)

print(df)
```

```python
result = df.loc["이영희", ["수학", "영어"]]

print(result)
```
