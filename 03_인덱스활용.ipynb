```python
import pandas as pd

dict_data = {
    "c0" : [1, 2, 3],
    "c1" : [4, 5, 6],
    "c2" : [7, 8, 9],
    "c3" : [10, 11, 12],
    "c4" : [13, 14, 15]
}

df = pd.DataFrame(dict_data, index=["r0", "r2", "r1"])
df
```

```python
# DataFrame객체.sort_index()
df.sort_index()
```

```python
ndf = df.sort_index(ascending = False)
ndf
```

```python
# 열 기준 정렬 (많이 쓰임)
# DataFrame객체.sort_values()
ndf = df.sort_values("c1",ascending = False)
ndf
```

```python
ndf2 = df.sort_values(by="c2",ascending=True)
ndf2
```

```python
ndf4 = df.sort_values(by=["c3", "c4"], ascending= [False, True])
ndf4
```
