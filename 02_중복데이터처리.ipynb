```python
import pandas as pd
```

```python
df.duplicated()
```

```python
df.duplicated(keep = "last")
```

```python
df.duplicated(keep=False)
```

```python
df["c2"].duplicated()
```

```python
df.duplicated(subset=["c2"])
```
