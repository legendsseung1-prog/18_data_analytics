```python
# 행/열 삭제
# 행삭제 : dataframe객체, drop(행 인덱스 또는 배열, axis = 0 Ehsms axis = "index")
# 열삭제 : dataframe객체, drop(열 이름 또는 배열, axis = 1 Ehsms axis = "columns")

import pandas as pd

exam_data = {
    "수학" : [90, 80, 70],
    "영어" : [98, 89, 95],
    "음악" : [ 85, 95, 100],
    "체육" : [100, 90, 90]
}

df = pd.DataFrame(exam_data, index = ["서준", "우현", "인아"])
df
```

```python
df2 = df.copy()
df2 = df2.drop("우현")
df2
```

```python
df3 = df.copy()
df3 = df3.drop(["우현", "인아"], axis = 0)
df3
```

```python
df4 = df.copy()
df4 = df4.drop(["우현", "인아"], axis = "index")
df4
```

```python
df5 = df.copy()
df5 = df5.drop(index = ["우현"])
df5
```

```python
# 열 삭제

df2 = df.copy()
df2.drop("수학", axis = 1)
```

```python
df3 = df.copy()
df3.drop(["영어", "음악"], axis = 1)
```

```python
df4 = df.copy()
df4.drop(["영어", "음악"], axis = "columns")
```

```python
df5 = df.copy()
df5 = df5.drop(columns = ["수학"])
df5
```

```python
# 행과 열의 위치 바꾸기
# DataFrame객체.transpose()
# DataFrame객체.T

exam_data = {
    "이름" : ["서준", "우현", "인아"],
    "수학" : [90, 80, 70],
    "영어" : [98, 89, 95],
    "음악" : [ 85, 95, 100],
    "체육" : [100, 90, 90]
}

import pandas as pd

df = pd.DataFrame(exam_data)
df
```

```python
df = df.transpose()
df
```

```python
df = df.T
df
```

```python
df = df.T
df
```
