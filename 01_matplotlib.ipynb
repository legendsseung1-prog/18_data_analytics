```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("./data/시도별_전출입_인구수.xlsx").ffill()
mask = (df["전출지별"] == "서울특별시") & (df["전입지별"] != "서울특별시")
df_seoul = df.loc[mask].drop(columns="전출지별").rename(columns={"전입지별": "전입지"})
sr_one = df_seoul.set_index("전입지").loc["경기도"]

plt.style.use("ggplot")
from matplotlib import font_manager
title_font = font_manager.FontProperties(fname="./data/malgun.ttf", size=30)
axis_font = font_manager.FontProperties(fname="./data/malgun.ttf", size=20)
legend_font = font_manager.FontProperties(fname="./data/malgun.ttf", size=25)

plt.figure(figsize=(14, 5))

plt.xticks(rotation="vertical")

plt.plot(sr_one.index, sr_one.values,
         marker="o", 
         markerfacecolor="red", 
         markeredgecolor="blue", 
         markeredgewidth=2, 
         markersize=10
         )

plt.title("서울 -> 경기 인구 이동", fontproperties=title_font)
plt.xlabel("기간", fontproperties=axis_font)
plt.ylabel("이동 인구수", fontproperties=axis_font)

plt.legend(labels=["서울 -> 경기"], loc="best", prop=legend_font)

plt.show()
```

```python
# 연도 눈금을 5년 간격으로 표시한 그래프
import matplotlib.pyplot as plt

years = list(sr_one.index)
positions = list(range(len(years)))
tick_positions = list(range(0, len(years), 5))
if tick_positions[-1] != len(years) - 1:
    tick_positions.append(len(years) - 1)

fig, ax = plt.subplots(figsize=(14, 5))
ax.plot(positions, sr_one.values)
ax.set_xticks(tick_positions, [years[i] for i in tick_positions])
ax.set_title("서울 -> 경기 인구 이동")
ax.set_xlabel("기간")
ax.set_ylabel("이동 인구수")
fig.tight_layout()
plt.show()
```
